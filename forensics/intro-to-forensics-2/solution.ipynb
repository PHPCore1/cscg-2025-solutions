```python
from scapy.utils import RawPcapReader
from scapy.layers.l2 import Ether, CookedLinux
from scapy.layers.inet import IP, TCP
import struct

def process_pcap(file_name):
    cap = ['']

    for (pkt_data, pkt_metadata,) in RawPcapReader(file_name):
        
        ether_pkt = CookedLinux(pkt_data)

        if 'pkttype' not in ether_pkt.fields:
            # LLC frames will have 'len' instead of 'type'.
            # We disregard those
            continue

        if ether_pkt.pkttype != 0x4:
            # disregard non-IPv4 packets
            continue

        ip_pkt = ether_pkt[IP]
        if ip_pkt.proto != 6:
            # Ignore non-TCP packet
            continue

        if ip_pkt.src != '192.168.178.21':
            continue

        tcp_pkt = ip_pkt[TCP]
        
        if tcp_pkt.dport == 80:
            continue

        data = struct.pack(">h", tcp_pkt.dport)
        
        if cap[-1] != data.decode():
            cap.append(data.decode())

    print(''.join(cap))
```

```python
process_pcap('./intro-forensics-2.pcapng')
```
