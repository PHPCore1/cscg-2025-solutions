```python
import requests
from lxml.etree import HTML, _Element, Comment
import re
import py_mini_racer
```

```python
BASE_URL = 'https://d8e7ce4bbe4313ae4330cf98-1024-intro-web-1.challenge.cscg.live:1337'
```

```python
res = requests.get(BASE_URL)
page_content = HTML(res.text)
```

```python
comment = list(page_content.xpath('/html/body/p[4]')[0].iter(Comment))[0]
flag_part_1 = re.findall(r'CSCG\{.+', comment.text)[0]
flag_part_1
```

```python
script = ''.join([''.join(t.itertext()) for t in list(page_content.iter('script'))])
script = script.replace('alert', 'return')
script += '\nshowFlag()'
script = re.sub(r'setInterval\(.+, \d+\)', 'null', script)
ctx = py_mini_racer.MiniRacer()
flag_part_2 = ctx.eval(script).replace("Here's the second part of your flag: ", '')
flag_part_2
```

```python
flag_part_3 = res.headers['FlagPart3']
flag_part_3
```

```python
flag = flag_part_1 + flag_part_2 + flag_part_3
print(flag)
```
