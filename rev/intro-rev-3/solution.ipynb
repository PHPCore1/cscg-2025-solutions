```python
STATIC_TEXT = 'b9yPw:MwqcoHuFz^r-o*{>I\x10Y'

def convert_chr_to_pwd(i, c):
    stage_1 = (ord(c) + 2) % 256
    original = (10 + i) ^ stage_1
    return chr(original)

PWD = ''.join([convert_chr_to_pwd(i, c) for i, c in enumerate(STATIC_TEXT)])
PWD
```

```python
session = remote('1b4556fb6aea8de07ed7b298-1024-intro-rev-3.challenge.cscg.live', 1337, ssl=True)
session.recvuntil(b'Give me your password: ')
session.sendline(PWD.encode())
session.recvuntil(b'Flag: ')
print(session.recvline(keepends=False).decode())
```
