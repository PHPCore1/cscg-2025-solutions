```python
import subprocess
import json
import string
import logging
import os
import re
import numpy as np
from z3 import *
```

```python
class VM:

    def __init__(self):
        self.stack = []
        self.reg = {}
        self.instruction_pointer = 0
        self.continue_breakpoint = False
        self.breakpoints = []
        self.known_flag = ''
        self.flag_addy_length = 0
        self.flag_addy_symbols = {}
        self.all_instructions_mapped = {
            'switch_first_and_second_on_stack': self.switch_first_and_second_on_stack,
            'end_program': self.end_program,
            'xor_registries_on_stack': self.xor_registries_on_stack,
            'push_reg_0_value_onto_stack': self.push_reg_0_value_onto_stack,
            'right_shift_stack': self.right_shift_stack,
            'decrement_reg_0_variable': self.decrement_reg_0_variable,
            'convert_stack_into_registry_and_clear': self.convert_stack_into_registry_and_clear,
            'clear_registry': self.clear_registry,
            'left_shift_stack': self.left_shift_stack,
            'increment_reg_0_variable': self.increment_reg_0_variable,
            'put_flag_chr_on_stack_with_reg_0_as_index': self.put_flag_chr_on_stack_with_reg_0_as_index,
            'jmp_by_reg_0_value': self.jmp_by_reg_0_value,
            'print_stack_as_unicode_and_clear': self.print_stack_as_unicode_and_clear,
            'nop': self.nop,
            'and_stack': self.and_stack,
            'convert_registry_to_stack_and_clear': self.convert_registry_to_stack_and_clear,
            'sub_stack': self.sub_stack,
            'add_stack': self.add_stack,
            'pop_stack_and_put_into_reg_0': self.pop_stack_and_put_into_reg_0,
        }

    def run(self):
        while self.instruction_pointer < len(program_instructions) and self.instruction_pointer >= 0:
            if (not self.continue_breakpoint) and self.instruction_pointer in self.breakpoints:
                print(f"Breakpoint hit at instruction {self.instruction_pointer}. Press Enter to continue...")
                self.continue_breakpoint = True
                return
            
            self.continue_breakpoint = False
            instruction = program_instructions[self.instruction_pointer]
            self.execute_instruction(instruction)

    def execute_instruction(self, instruction):
        self.all_instructions_mapped[instruction]()

    def get_flag_chr(self, index):
        if index < len(self.known_flag):
            return BitVecVal(ord(self.known_flag[index]), 64)
        elif index < len(self.known_flag) + self.flag_addy_length:
            offset = index - len(self.known_flag)
            self.flag_addy_symbols[offset] = self.flag_addy_symbols.get(offset, BitVec("FLAG_ADDY_" + str(offset), 64))
            return self.flag_addy_symbols[offset]
        else:
            return BitVecVal(-1, 64)

    def get_registry_value(self, reg):
        return self.reg.get(reg, BitVecVal(0, 64))

    def exec_function_on_by_stack_referenced_operands(self, func):
        top = simplify(self.stack.pop(0))
        second = simplify(self.stack.pop(0))
        self.reg[top.as_signed_long()] = func(self.get_registry_value(top.as_signed_long()), self.get_registry_value(second.as_signed_long()))

    def switch_first_and_second_on_stack(self):
        self.stack[0], self.stack[1] = self.stack[1], self.stack[0]
        self.instruction_pointer += 1

    def end_program(self):
        self.instruction_pointer = len(program_instructions)

    def xor_registries_on_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x ^ y)
        self.instruction_pointer += 1

    def push_reg_0_value_onto_stack(self):
        self.stack.insert(0, self.get_registry_value(0))
        self.instruction_pointer += 1

    def right_shift_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x >> y)
        self.instruction_pointer += 1

    def decrement_reg_0_variable(self):
        self.reg[0] = simplify(self.get_registry_value(0) - BitVecVal(1, 64))
        self.instruction_pointer += 1

    def convert_stack_into_registry_and_clear(self):
        self.reg = dict(enumerate(self.stack))
        self.stack = []
        self.instruction_pointer += 1

    def clear_registry(self):
        self.reg = {}
        self.instruction_pointer += 1

    def left_shift_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x << y)
        self.instruction_pointer += 1

    def increment_reg_0_variable(self):
        self.reg[0] = simplify(self.get_registry_value(0) + BitVecVal(1, 64))
        self.instruction_pointer += 1

    def put_flag_chr_on_stack_with_reg_0_as_index(self):
        flag_ascii = self.get_flag_chr(simplify(self.get_registry_value(0)).as_signed_long())
        self.stack.insert(0, flag_ascii)
        self.instruction_pointer += 1

    def jmp_by_reg_0_value(self):
        self.instruction_pointer += simplify(self.get_registry_value(0)).as_signed_long()

    def print_stack_as_unicode_and_clear(self):
        print("".join(map(lambda x: chr(simplify(x).as_signed_long()), self.stack)))
        self.stack = []
        self.instruction_pointer += 1

    def nop(self):
        self.instruction_pointer += 1

    def and_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x & y)
        self.instruction_pointer += 1

    def convert_registry_to_stack_and_clear(self):
        self.stack = list(map(lambda x: x[1], sorted(self.reg.items(), key=lambda x: x[0])))
        self.reg = {}
        self.instruction_pointer += 1

    def sub_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x - y)
        self.instruction_pointer += 1

    def add_stack(self):
        self.exec_function_on_by_stack_referenced_operands(lambda x, y: x + y)
        self.instruction_pointer += 1

    def pop_stack_and_put_into_reg_0(self):
        self.reg[0] = self.stack.pop(0)
        self.instruction_pointer += 1
```

```python
def get_test(pwd):
    data = subprocess.check_output(["bash", "test.sh", pwd], stderr=subprocess.STDOUT)
    lines = data.split(b'\n')
    lines = lines[:-2]

    state_data = []

    for line in lines:
        if line.startswith(b'trace: Data: '):
            state_data.append(json.loads(line[13:].decode()))
            state_data[-1].pop('transformed_flag')
        else:
            meta_data = state_data[-1].get('meta', [])
            meta_data.append(line[7:].decode())
            state_data[-1]['meta'] = meta_data

    #state = list(filter(lambda x: 'meta' in x and any(map(lambda y: 'Jumping' in y, x['meta'])), state_data))[-2]
    #reg = state['registry']
    #registry_list = list(map(lambda x: x[1], sorted(reg.items(), key=lambda x: int(x[0]))))
    return state_data#registry_list[10:19][::-1], registry_list[19:][::-1]

def cmp(res):
    should_be = ''
    is_currently = ''
    for cmp_to, _input in zip(res[0], res[1]):
        should_be += bin(cmp_to)[2:].zfill(32)
        is_currently += bin(_input)[2:].zfill(32)

    return should_be, is_currently

def jmp_analyzer(res):
    jmp_instructions = list(filter(lambda x: 'meta' in x and any(map(lambda y: 'Jumping' in y, x['meta'])), res))
    jmps = list(map(lambda x: tuple(map(int, re.findall(r"Jumping at (\d+) to (\d+)", x['meta'][0])[0])), jmp_instructions))
    jmp_counter = {}
    loop_counter = {}

    for jmp in jmps:
        jmp_counter[jmp] = jmp_counter.get(jmp, 0) + 1

        if jmp[0] < jmp[1]:
            continue
        loop_counter[jmp] = loop_counter.get(jmp, 0) + 1

    return jmp_instructions, jmp_counter, loop_counter
```

```python
vm = VM()
vm.known_flag = 'dach20'
vm.flag_addy_length = 2

total_flag_length = len(vm.known_flag) + vm.flag_addy_length

vm.reg = {
    0: BitVecVal(1, 64), 
    1: BitVecVal(4, 64), 
    2: BitVecVal(0, 64), 
    3: BitVecVal(total_flag_length, 64), 
    4: BitVecVal(1, 64), 
    5: BitVecVal(-306, 64)
}
vm.stack = [vm.reg[3]]
vm.breakpoints = [52205]
vm.instruction_pointer = 309
vm.run()
```

```python
high_reg = max(vm.reg.keys())
target_chunks = (total_flag_length + 7) // 8 * 2
start_calculated = high_reg - target_chunks + 1
start_target = high_reg - 2 * target_chunks + 1
chunks_compare_length = 2
```

```python
solver = Solver()
```

```python
goal = Goal()
for i in range(chunks_compare_length):
    goal.add(
        vm.reg[start_calculated + i] == vm.reg[start_target + i]
    )

for sym in vm.flag_addy_symbols.values():
    goal.add(sym & ~0x7F == 0)
```

```python
set_param('sat.euf', True) 
set_param('smt.ematching', False)  

for g in Tactic('simplify')(goal):
    solver.add(g)
```

```python
solver.check()
```
