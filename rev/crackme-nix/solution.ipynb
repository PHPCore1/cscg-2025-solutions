```python
state_data = []

for line in lines:
    if line.startswith(b'trace: Data: '):
        state_data.append(json.loads(line[13:].decode()))
        state_data[-1]['instruction'] = program_instructions[state_data[-1]['instruction_pointer']]
        state_data[-1]['prev_instruction'] = state_data[-2]['instruction'] if len(state_data) > 1 else ''
        state_data[-1].pop('transformed_flag')
    else:
        meta_data = state_data[-1].get('meta', [])
        meta_data.append(line[7:].decode())
        state_data[-1]['meta'] = meta_data
```

```python
state_data_with_ascii = []

for state in state_data:
    deep_clone = json.loads(json.dumps(state))
    deep_clone['stack'] = list(map(lambda x: chr(x) if x in range(0x110000) else x, deep_clone['stack']))
    
    for key in deep_clone['registry']:
        try:
            deep_clone['registry'][key] = chr(deep_clone['registry'][key])
        except:
            pass

    state_data_with_ascii.append(deep_clone)
```
