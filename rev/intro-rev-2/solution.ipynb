```python
from pwn import *
```

```python
STATIC_TEXT = b'\x02\xea\x02\xe8\xfc\xfd\xbd\xfd\xf2\xec\xe8\xfd\xfb\xea\xf7\xfc\xef\xb9\xfb\xf6\xea\xfd\xf2\xf8\xf7\x00'

def convert_chr_to_pwd(c):
    return chr((c + 0x77) % 256)

PWD = ''.join([convert_chr_to_pwd(c) for c in STATIC_TEXT])
PWD
```

```python
session = remote('3da300cb3c25fb362c25b698-1024-intro-rev-2.challenge.cscg.live', 1337, ssl=True)
session.recvuntil(b'Give me your password: ')
session.sendline(PWD.encode())
session.recvuntil(b'Flag: ')
print(session.recvline(keepends=False).decode())
```
