```python
from pwn import *
```

```python
def xor(a, b):
    return bytes([x ^ y for x, y in zip(a, b)])
```

```python
TEST_STRING = "A" * 96
```

```python
session = remote('33b06ceccd4a6496f6be8f95-1024-intro-crypto-1.challenge.cscg.live', 1337, ssl=True)
```

```python
session.recvuntil(b'Enter some plaintext (hex): ')
session.sendline(TEST_STRING.encode().hex().encode())
```

```python
encrypted_test_strings = []

for i in range(0, 255):
    encrypted_test_strings.append(bytes.fromhex(session.recvline(keepends=False).split(b': ')[1].decode()))
```

```python
session.recvuntil(b'Flag:')
encrypted_flag = bytes.fromhex(session.recvline(keepends=False).decode())
```

```python
session.close()
```

```python
# Key-Stream Duplication in AES-CTR Mode because of Nonce-Reuse
for encrypted_test_string in encrypted_test_strings:
    keystream = xor(encrypted_test_string, TEST_STRING.encode())
    decrypted_flag = xor(encrypted_flag, keystream)
    if b'CSCG{' in decrypted_flag:
        print(decrypted_flag)
        break
```
