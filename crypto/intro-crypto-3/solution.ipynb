```python
import z3
```

```python
data = []

with open('./msg.txt') as f:
    for i, line in enumerate(f):
        if i == 0 or len(line.strip()) == 0:
            continue

        data.append(int(line.strip()))
```

```python
FLAG_LEN = len(data)

known_data = [
    (0, ord('C')),
    (1, ord('S')),
    (2, ord('C')),
    (3, ord('G')),
    (4, ord('{')),
    (FLAG_LEN - 1, ord('}')),
]

BITS = 56
MASK = 0xFF

def rng(x, size, a, b):
    return (x*a+b) & ((2**size)-1)
```

```python
flag_symbols = [z3.BitVec(f'flag_{i}', BITS) for i in range(FLAG_LEN)]
state_symbols = [z3.BitVec(f'state_{i}', BITS) for i in range(FLAG_LEN + 1)]
output_symbols = [z3.BitVec(f'output_{i}', BITS) for i in range(FLAG_LEN)]
A = z3.BitVec('A', BITS)
B = z3.BitVec('B', BITS)
```

```python
known_flag_equations = [flag_symbols[known_index] == known_value for known_index, known_value in known_data]
known_output_equations = [output_symbols[known_index] == known_value for known_index, known_value in enumerate(data)]
output_equations = [(flag_symbols[i] ^ (state_symbols[i + 1] & MASK)) == output_symbols[i] for i in range(FLAG_LEN)]
state_equations = [rng(state_symbols[i - 1], BITS, A, B) == state_symbols[i] for i in range(1, FLAG_LEN + 1)]
all_equations = [*known_flag_equations, *known_output_equations, *output_equations, *state_equations]
all_equations
```

```python
s = z3.Solver()
s.add(*all_equations)
s.check()
model = s.model()
```

```python
flag = ''.join([chr(model[flag_symbol].as_long()) for flag_symbol in flag_symbols])
print(flag)
```
